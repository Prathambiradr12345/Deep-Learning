```python
from keras.applications.vgg16 import VGG16
#load model
model = VGG16()
#summarize the model
model.summary()
```

```python
from keras.utils import plot_model
plot_model(model, to_file='vgg.png')
```

```python
for i in range(len(model.layers)):
    if 'conv' not in model.layers[i].name:#check for convolutional layers
        continue
    filters, biases = model.layers[i].get_weights()#get filters weights
    print(model.layers[i].name, filters.shape)
```

```python
# retrieve second hidden layer
filters,bias=model.layers[1].get_weights()
print(filters.shape)
```

```python
f_min, f_max = filters.min(), filters.max()#normalization
filters = (filters - f_min) / (f_max - f_min)
```

```python
import matplotlib.pyplot as pyplot

n_filters = 6
ix = 1
fig = pyplot.figure(figsize=(15, 10))
for i in range(n_filters):
    f = filters[:, :, :, i]
    for j in range(3):
        pyplot.subplot(n_filters, 3, ix)
        pyplot.imshow(f[:, :, j], cmap='gray')
        ix += 1
pyplot.show()
```

```python
from keras.models import Model # import the Model class
from keras.applications.vgg16 import VGG16

# ... (rest of your existing code)

model=Model(inputs=model.inputs,outputs=model.layers[1].output)
```

```python
from keras.preprocessing.image import load_img, img_to_array

image = load_img('/content/cat.jpg', target_size=(224, 224))
image = img_to_array(image)
image = image.reshape((1, image.shape[0], image.shape[1], image.shape[2]))
```

```python
#calculating features
features = model.predict(image)

# Iterate through the feature maps and display them separately
import matplotlib.pyplot as plt

for i, feature_map in enumerate(features):
    fig = plt.figure(figsize=(20, 15))
    # Calculate the number of filters in the feature map
    num_filters = feature_map.shape[3]
    # Calculate the grid dimensions (rows and columns)
    grid_size = int(num_filters**0.5) + 1  # Ensure enough space
    for j in range(num_filters):  # Iterate through filters
        # Calculate subplot index based on the grid
        subplot_index = j + 1
        plt.subplot(grid_size, grid_size, subplot_index)  #Dynamic grid size
        plt.imshow(feature_map[0, :, :, j], cmap='gray')
    plt.show()
```

```python
model=VGG16()
layer_index=[2,5,9,13,17]
outputs=[model.layers[i].output for i in layer_index]
model=Model(inputs=model.inputs,outputs=outputs)
```

```python
feature_map=model.predict(image)
for i ,fmap in zip(layer_index,feature_map):
    fig=pyplot.figure(figsize=(20,15))
    # Calculate the number of filters in the feature map
    num_filters = fmap.shape[3]
    # Calculate the grid dimensions (rows and columns)
    grid_size = int(num_filters**0.5) + 1  # Ensure enough space
    for j in range(num_filters):  # Iterate through filters
        # Calculate subplot index based on the grid
        subplot_index = j + 1
        pyplot.subplot(grid_size, grid_size, subplot_index)  #Dynamic grid size
        pyplot.imshow(fmap[0,:,:,j],cmap='gray')
    pyplot.show()
```
