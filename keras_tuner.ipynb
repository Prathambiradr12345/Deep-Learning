```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('diabetes.csv')
```

```python
df
```

```python
df.corr()['Outcome']
```

```python
x=df.iloc[:,:-1].values
y=df.iloc[:,-1].values
```

```python
x
```

```python
y
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
x=scaler.fit_transform(x)
```

```python
x.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
import tensorflow as tf
from tensorflow import keras
```

```python
from keras import Sequential
from keras.layers import Dense
```

```python
model=Sequential()
model.add(Dense(32,activation='relu',input_dim=8))
model.add(Dense(1,activation='sigmoid'))
model.compile(optimizer='Adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model.fit(batch_size=32,x=x_train,y=y_train,epochs=100,validation_data=(x_test,y_test))
```

```python
import keras_tuner as kt
```

```python
def build_model(hp):
    model=Sequential()
    model.add(Dense(32,activation='relu',input_dim=8))
    model.add(Dense(1,activation='sigmoid'))
    model.compile(optimizer=hp.Choice('optimizer',values=['Adam','sgd','rmsprop','adadelta']),loss='binary_crossentropy',metrics=['accuracy'])
    return model
```

```python
tuner=kt.RandomSearch(build_model,objective='val_accuracy',max_trials=5)
```

```python
tuner.search(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
model=tuner.get_best_models(num_models=1)[0]
```

```python
model.summary()
```

```python
model.fit(x_train,y_train,batch_size=32,epochs=100,initial_epoch=6,validation_data=(x_test,y_test))
```

```python
def build_model(hp):
    model = Sequential()

    c = 0
    for i in range(hp.Int('num_layers', min_value=1, max_value=10)):
        if c == 0:
            # Corrected line: Change 'strp' to 'step'
            model.add(Dense(hp.Int('units' + str(i), min_value=8, max_value=120, step=8),
                            activation=hp.Choice('activation' + str(i), values=['relu', 'tanh', 'sigmoid']),
                            input_dim=8))
        else:
            # Corrected line: Change 'strp' to 'step'
            model.add(Dense(hp.Int('units' + str(i), min_value=8, max_value=120, step=8),
                            activation=hp.Choice('activation' + str(i), values=['relu', 'tanh', 'sigmoid'])))
        c += 1
    # Corrected line: Moved outside the loop to add the output layer only once
    model.add(Dense(1, activation='sigmoid'))
    model.compile(optimizer=hp.Choice('optimizer', values=['Adam', 'sgd', 'rmsprop', 'adadelta']),
                  loss='binary_crossentropy',
                  metrics=['accuracy'])
    return model
```
